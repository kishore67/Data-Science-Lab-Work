```python
### ZIP FUNCTION:
      ## initializing lists
name=["kishore","mugesh","dharunima"]
roll_no=[1,2,3]
score=[70,71,72]
todays_score=[35,36,37]

## use zip() to map values

mapped=zip(name,roll_no,score,todays_score)
mapped=list(mapped)

print(mapped)
type(mapped)
```
