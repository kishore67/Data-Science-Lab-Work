# Fun Level - Hand Cricket

“This project simulates a complete hand cricket match with a head-tail toss and target-based scoring, designed with structured output and user interaction for a realistic gameplay experience.”

```python
import random

# -----------------------------
# Toss Function
# -----------------------------
def toss():
    print("\n" + "="*40)
    print("TOSS - HEAD / TAIL")
    print("="*40)

    print("1. Head")
    print("2. Tail")

    while True:
        try:
            user = int(input("\nChoose (1/2): "))
            if user not in [1, 2]:
                print("Invalid choice. Try again.")
                continue
            break
        except:
            print("Enter a valid number.")

    system = random.randint(1, 2)
    choices = {1: "Head", 2: "Tail"}

    print("\n" + "-"*40)
    print(f"Toss Result   : {choices[system]}")
    print(f"Your Choice   : {choices[user]}")
    print("-"*40)

    if user == system:
        print("You won the toss.")
        return "user"
    else:
        print("System won the toss.")
        return "system"


# -----------------------------
# Innings Function
# -----------------------------
def play_innings(player, target=None):
    score = 0

    print("\n" + "="*40)
    print("NEW INNINGS STARTED")
    print("="*40)

    if target is not None:
        print(f"Target: {target + 1}")

    while True:
        try:
            user = int(input("\nEnter your run (1-6): "))
            if user < 1 or user > 6:
                print("Enter a number between 1 and 6.")
                continue
        except:
            print("Invalid input.")
            continue

        system = random.randint(1, 6)

        print("\n" + "-"*40)
        print(f"You     : {user}")
        print(f"System  : {system}")
        print("-"*40)

        if user == system:
            print("Out. Innings over.")
            break

        if player == "user":
            score += user
        else:
            score += system

        print(f"Current Score: {score}")

        if target is not None and score > target:
            print("Target achieved.")
            return score

    return score


# -----------------------------
# One Full Match
# -----------------------------
def play_match():
    print("\nHAND CRICKET GAME")

    winner = toss()

    # Decision after toss
    if winner == "user":
        print("\nChoose:")
        print("1. Bat")
        print("2. Bowl")

        while True:
            choice = input("Enter 1 or 2: ")
            if choice == "1":
                choice = "bat"
                break
            elif choice == "2":
                choice = "bowl"
                break
            else:
                print("Invalid choice.")

        if choice == "bat":
            print("\nYou are batting first.")
            user_score = play_innings("user")

            print(f"\nTarget for system: {user_score + 1}")

            print("\nSystem batting...")
            system_score = play_innings("system", user_score)

        else:
            print("\nYou are bowling first.")
            system_score = play_innings("system")

            print(f"\nTarget for you: {system_score + 1}")

            print("\nYou batting...")
            user_score = play_innings("user", system_score)

    else:
        decision = random.choice(["bat", "bowl"])

        if decision == "bat":
            print("\nSystem chose to bat first.")
            system_score = play_innings("system")

            print(f"\nTarget for you: {system_score + 1}")

            print("\nYou batting...")
            user_score = play_innings("user", system_score)

        else:
            print("\nSystem chose to bowl first.")
            print("\nYou batting first.")
            user_score = play_innings("user")

            print(f"\nTarget for system: {user_score + 1}")

            print("\nSystem batting...")
            system_score = play_innings("system", user_score)

    # Final Result
    print("\n" + "="*40)
    print("FINAL MATCH RESULT")
    print("="*40)

    print(f"Your Score   : {user_score}")
    print(f"System Score : {system_score}")
    print("-"*40)

    if user_score > system_score:
        print("Result: You win.")
    elif user_score < system_score:
        print("Result: System wins.")
    else:
        print("Result: Match tie.")

    print("="*40)


# -----------------------------
# Main Function (Replay)
# -----------------------------
def main():
    while True:
        play_match()

        again = input("\nDo you want to play again? (yes/no): ").lower()
        if again != "yes":
            print("\nThank you for playing.")
            break


# -----------------------------
# Run Game
# -----------------------------
if __name__ == "__main__":
    main()
```
