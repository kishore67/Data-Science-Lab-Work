Container Types in Python

The four core built-in container types are:

List:
A mutable, ordered sequence of elements.
Lists are indexed by integers and can hold items of different data types.
They are defined using square brackets.
Example:
"[1, "a", 2.5]"

Tuple:
An immutable, ordered sequence of elements.
Tuples are similar to lists but cannot be changed after creation.
They are defined using parentheses.
Example:
"(1, "a", 2.5)"

Dictionary (dict):
A mutable mapping type that stores data as key–value pairs.
Data is accessed using unique keys, not index positions.
Dictionaries are defined using curly braces.
Example:
"{"name": "John", "age": 30}"

Set:
A mutable, unordered collection of unique elements.
Sets are useful for operations like union, intersection, and membership testing.
They are defined using curly braces or the "set()" function.
Example:
"{1, 2, 3}"

Note:
Strings ("str") are also considered a sequence container type because they are ordered.

```python
## Python as a Calculator
## Given;
## TP=50
## FP=8
## TN=32
## FN=10

TP=50
FP=8
TN=32
FN=10

total=TP+FP+TN+FN
print("Total predictions : ",total)

#Accuracy
accuracy=(TP+TN)/total
print("Accuracy : ",accuracy)

#Precision
precision=TP/(TP+FP)
print("Precision : ",precision)

#Recall
recall=TP/(TP+FN)
print("Recall : ",recall)

#F1_score
f1_score=2*precision*recall/precision+recall
print("F1_Score : ",f1_score)
```

```python
## R²(Coefficient of Determination)
## Given;
## SSR=420
## SSE=180

SSE=420
SSR=180

## Total sum of Squares
SST=SSE+SSR
print("Total sum of squares : ",SST)

##R^2 VALUE
R_Square=SSR/SST
print(R_Square)

##Interpretation

if R_Square>=0.7:
  print("Model explains 70% or more varience")
else:
 print("Model explains < 70% varience")
```

```python
## Mean Squared Error
## Given
## 4,9,16,1,25

## Store Squared Errors
error=[4,9,16,1,25]

## Number of observation
n=len(error)

## Mean Squared Error
mse=sum(error)/n
print("Mean Squared Error : ",mse)
```

```python
## Cooks Distance

## Given;
# e_i^2 = 9
# MSE = 3
# p = 3
# h_ii = 0.2
# n = 25

# Formula Used = D_i= e_i^2       h_ii
            #        -------  * ---------
               #    (p+1)*MSE   (1-h_ii)^2


# Assign values
e2 = 9
MSE = 3
p = 3
h = 0.2
n = 25

# Cook's Distance formula
cook = (e2 / ((p + 1) * MSE)) * (h / ((1 - h) ** 2))
print("Cook's Distance =", cook)

# Threshold
threshold = 4 / n
print("Threshold =", threshold)

# Decision
if cook > threshold:
    print("Observation is influential")
else:
    print("Observation is NOT influential")
```
